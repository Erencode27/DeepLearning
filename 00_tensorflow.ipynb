It looks like if we want our shuffled tensors to be in the same order, we've 
got to use the global level random seed as well as the operation level random seed:
Rule:
If both the global and the operation seed are set: Both seeds are used in conjunction to determine the random sequence.

 # MATRİX MULTIPLICATION
 In machine learning, matrix multiplication is one of the most common tensor operations.


```python
# Matrix multiplication

tf.linalg.matmul(tensor5,tensor6)
```

```python
# Matrix multiplication with Python operator "@"
tensor5 @ tensor6
```

```python
tensor5.shape
tensor6.shape
```

```python
# Create a tensor (3,2)
X1 = np.arange(1,7)
X1
tensor7 =tf.constant(X1,shape=[2,3])
# Create another (3,2) tensor
Y1 = np.arange(7,13)
tensor8 = tf.constant(Y1,shape=[3,2])
```

```python
tensor7, tensor8
```

```python
tf.linalg.matmul(tensor7,tensor8)
```

**! The number of columns of the 1st matrix must equal the number of rows of the 2nd matrix!**
* İlk matrix'in sütun sayısı ile 2. matrix'in satır sayısının eşit olması lazım matrix multiplication yapılabilmesi için

### There are two rules our tensors (or matrices) need to fulfill if we're going to matrix multiply them:
1. the inner dimensions must match
2. the resulting matrix has the shape of the inner dimensions

```python
# İncompatiable olan tensorlerin şeklini değiştirebilmek için ayrıca tf.reshape()'de kullanılabilir;
tf.reshape(tensor7,shape=(3,2))
tensor7, tensor8
```

```python
# Try to matrix multiply X by reshape Y 
tensor8 @ tf.reshape(Y1,shape=(2,3))
```

```python
tensor7,tf.transpose(tensor7)
```

```python
tensor7,tensor8
```

```python
tf.reshape(tensor7,shape=(3,2))
tensor7
tf.matmul(tensor7,tensor8)
```

#### **The Dot Product** 
Matrix multiplication is also referred to as the dot product.
You can perform matrix multiplication using:
* tf.linalg.matmul()
* tf.tensordot()
  

```python
# Perform the dot product on tensor7 and tensor8 (requires X or Y to be transposed)
tf.tensordot(tensor7,tensor8,axes= 1)
```

```python
X = tf.constant([[1, 2],
                 [3, 4],
                 [5, 6]])

# Create another (3, 2) tensor
Y = tf.constant([[7, 8],
                 [9, 10],
                 [11, 12]])
X, Y
```

```python
# Example of reshape (3, 2) -> (2, 3)
tf.reshape(Y, shape=(2, 3))
```

```python
# Try matrix multiplication with reshaped Y
X @ tf.reshape(Y, shape=(2, 3))
```

```python
# Example of transpose (3, 2) -> (2, 3)
tf.transpose(X)
```

```python
#Try matrix multiplication 
tf.matmul(tf.transpose(X), Y)
```

```python
# Perform the doc product on X and Y (requires X or Y to be transposed)
tf.tensordot(tf.transpose(X),Y,axes=1)
```

```python
# Perform matrix multiplication between X and Y (transposed)
tf.matmul(X,tf.transpose(Y))
```

```python
# Perform matrix multiplication between X and Y (reshaped)
tf.matmul(X,tf.reshape(Y,shape =(2,3)))
```

```python
# Check the values of Y, reshape Y and transposed Y
print("Normal Y:",Y)
print("Y reshaped to (2,3)",tf.reshape(Y,shape=(2,3)))
print("Y transposed",tf.transpose(Y))
```

**Generally,when performing matrix multiplication on two tensors and one of the axes doesn't line up, you will transpose(rather than reshape) one of the tensors to satisfy matrix multiplication rules.**
